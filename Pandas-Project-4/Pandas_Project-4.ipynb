```python
## Convert the datatypes of START_DATE and END_DATE columns to datetime
df['START_DATE'] = pd.to_datetime(df['START_DATE'])
df["END_DATE"] = pd.to_datetime(df["END_DATE"])
df.dtypes
```

```python
## Extract the month from START_DATE and try to get the proportion of rides of different months.
df['MONTH'] = df['START_DATE'].dt.month
df['MONTH_NAME'] = df['START_DATE'].dt.month_name()
df.head()
```

```python
df.groupby("MONTH")['MILES '].sum()
```

```python
ride_counts = df[['MONTH','MONTH_NAME']].value_counts().sort_index()
print(ride_counts)
```

```python
proportion = (ride_counts/ride_counts.sum()) * 100
proportion = proportion.round(2)
print(proportion)
```

```python
## Find the average distance covered each month
avg_dist = df.groupby(["MONTH","MONTH_NAME"])['MILES '].mean()
avg_dist = avg_dist.round(2)
print(avg_dist)
```

```python
## Extract the day from the START_DATE column
df["DAY"] = df['START_DATE'].dt.day
df.head()
```

```python
## Find the total miles covered per category per purpose
df.groupby(["CATEGORY ","PURPOSE "]).sum(numeric_only=True)["MILES "].sort_values(ascending=False).reset_index(name="TOTAL_MILES")
```
