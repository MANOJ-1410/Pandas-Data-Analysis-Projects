Seasonality

```python
## Seasonality = predictable changes that repeat over a fixed period of time.
## Examples : Traffic increases every morning & evening
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import month_plot, quarter_plot
```

```python
# Plotting the monthly seasonality
month_plot(df['Close'].resample('M').mean())
plt.show()
```

```python
## By looking graphs remember this
# Black Lines : These are the actual monthly series (each year has a spike).
# That means your dataset has very strong seasonal cycles every month.
# Red Line : This is the average value of each month across all years.
```

```python
# Plotting the quarterly seasonality
quarter_plot(df['Close'].resample('Q').mean())
plt.show()
```

```python
# Seasonal Decomposition Plots for Bitcoin Data
decomposition = seasonal_decompose(df['Close'], model='add', period=365)
fig = decomposition.plot()
```

✅ Additive model
                    Observed = Trend + Seasonal + Residual
👉 Seasonal effect is constant size over time.
👉 Good when your data fluctuates by the same amount regardless of the level.

Example:
Temperature ,
Daily website traffic (stable scale) , 
Sales where seasonal effect is roughly constant

✅ Multiplicative model
                        Observed = Trend × Seasonal × Residual
👉 Seasonal effect grows or shrinks with the level of the series.
👉 Good when your data fluctuates by a percentage.

Example:
Stock prices (fluctuations increase as price increases) ,
Marketing spend vs revenue ,
Sales with holiday spikes

Explaination for the above diagram : Decomposition split into 4 pieces,
🔹 Original Series (Close)
🔹 Trend = long-term direction
🔹 Seasonal = repeated yearly pattern
🔹 Residual = randomness



```python
# Seasonal Decomposition Plots for Chocolate Revenue Data
Decomposition1 = seasonal_decompose(df_second['revenue'],model='mul',period=12)
fig = Decomposition1.plot()
fig.set_size_inches(10,8)
plt.show()
```

Auto Correlation Function (ACF)

```python
from statsmodels.graphics.tsaplots import plot_acf

# Plot the autocorrelation (ACF)
fig, ax = plt.subplots(figsize=(10,6))
plot_acf(df['Close'], lags=100, ax=ax)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,6))
plot_acf(df_second['revenue'], lags=50, ax=ax)
plt.show()
```

Partial Auto relation Function (PACF)

PACF tells direct connection between data points at various times without the effect of in between points

```python
from statsmodels.graphics.tsaplots import plot_pacf

```

```python
fig, ax = plt.subplots(figsize = (10,6))
plot_pacf(df['Close'], lags=100 , ax=ax)
plt.show()
```

```python
fig , ax = plt.subplots(figsize = (10,6))
plot_pacf(df_second['revenue'], lags=30, ax=ax)
plt.show()
```
